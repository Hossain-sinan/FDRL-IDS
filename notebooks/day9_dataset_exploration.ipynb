```python
file_path = r"D:\Projects\FDRL-IDS\data\raw\Tuesday-WorkingHours.pcap_ISCX.csv"

df_tuesday = pd.read_csv(file_path)

print("Shape:", df_tuesday.shape)
```

```python
print(df_tuesday.columns.tolist())
```

```python
import numpy as np

print("=" * 60)
print("TUESDAY DATASET EXPLORATION")
print("=" * 60)

# 1. Shape
print("\n1. SHAPE")
print("Rows:", df_tuesday.shape[0])
print("Columns:", df_tuesday.shape[1])

# 2. Data types
print("\n2. DATA TYPES")
print(df_tuesday.dtypes.value_counts())

# 3. Labels
print("\n3. LABELS")
print(df_tuesday[" Label"].value_counts())

# 4. Duplicate columns
print("\n4. DUPLICATE COLUMNS")
duplicate_columns = df_tuesday.columns[df_tuesday.columns.duplicated()]
print(duplicate_columns.tolist())

# 5. Constant columns
print("\n5. CONSTANT COLUMNS")
constant_columns = df_tuesday.columns[df_tuesday.nunique() <= 1]
print(constant_columns.tolist())

# 6. Missing values
print("\n6. MISSING VALUES")
missing = df_tuesday.isna().sum()
print("Total:", missing.sum())
print(missing[missing > 0])

# 7. Infinite values
print("\n7. INFINITE VALUES")
numeric_df = df_tuesday.select_dtypes(include=np.number)
infinite = np.isinf(numeric_df).sum()
print("Total:", infinite.sum())
print(infinite[infinite > 0])

# 8. Suspicious minimum values
print("\n8. MINIMUM VALUES")
print(df_tuesday.describe().T["min"].sort_values().head(10))

# 9. Feature statistics
print("\n9. FEATURE STATISTICS")
print(df_tuesday.describe().T)

# 10. Unique labels
print("\n10. UNIQUE LABELS")
print(df_tuesday[" Label"].unique())

print("\n" + "=" * 60)
print("TUESDAY INSPECTION COMPLETE")
print("=" * 60)
```

```python
print("=== TUESDAY SUMMARY ===")

print("\nShape:")
print(df_tuesday.shape)

print("\nData types:")
print(df_tuesday.dtypes.value_counts())

print("\nLabels:")
print(df_tuesday[" Label"].value_counts())

print("\nDuplicate columns:")
print(df_tuesday.columns[df_tuesday.columns.duplicated()].tolist())

print("\nConstant columns:")
print(df_tuesday.columns[df_tuesday.nunique() <= 1].tolist())

missing = df_tuesday.isna().sum()
print("\nMissing values:")
print("Total:", missing.sum())
print(missing[missing > 0])

numeric_df = df_tuesday.select_dtypes(include=np.number)
infinite = np.isinf(numeric_df).sum()

print("\nInfinite values:")
print("Total:", infinite.sum())
print(infinite[infinite > 0])

print("\nSuspicious minimum values:")
print(df_tuesday.describe().T["min"].sort_values().head(10))
```
